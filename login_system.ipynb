```python
num1 = float(input("Add first number:"))
operator = input("Enter operator +,-,*,/:")
num2 =  float(input("Add second number:"))
if operator == "+":
      print("Result:", num1-num2)

elif operator == "-":
        print("Result:,num1 * num2")
elif operator =="/":
    if num2 !=0:
      print("Result:", num1 / num2)
else :
     print("Enter wrong operator! only +, -, *, / use")
```
